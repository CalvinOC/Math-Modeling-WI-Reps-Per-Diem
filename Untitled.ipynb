```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fsolve
from scipy.optimize import curve_fit
%matplotlib inline
#import pandas as pd
```

```python
#total money, days claimed, district
#distance from madison, district
import csv
repsData = []
with open ('WI_reps_perdiem_data.csv', newline='') as f:
    reader = csv.reader(f)
    header = next(reader) #adds headers to columns(?)
    print([header[0],header[2],header[3],header[4]])
    for row in reader:
        repsData.append([row[0],row[2],row[3],row[4]])
del repsData[-1] #removes the total line at the bottom

hometownData = []
with open ('Wisconsin_Assembly_2025_26_Committee_and_Hometown_Data.csv', newline='') as f:
    reader = csv.reader(f)
    header = next(reader) #adds headers to columns(?)
    print([header[0],header[1],header[4],header[5]])
    for row in reader:
        hometownData.append([row[0],row[1],row[4],row[5]])
del hometownData[34] #remove the person who lives in madison and didnt spend any money

#repsData = pd.read_csv(r'C:\Users\reehd0082\Downloads\WI_reps_perdiem_data.csv', delimiter=',',encoding='windows-1252')
#homeData = pd.read_csv(r'C:\Users\reehd0082\Downloads\Wisconsin_Assembly_2025_26_Committee_and_Hometown_Data.xlsx', delimiter=',',usecols=[0,4,5],encoding='windows-1252')
#assume party has no impact
```

```python
for row in repsData:
    row[1] = int(row[1])
repsDataSorted = sorted(repsData, key=lambda x: x[1])
```

```python
for row in hometownData:
    row[0] = int(row[0])
hometownDataSorted = sorted(hometownData, key=lambda x: x[0])
```

```python
print(len(repsDataSorted))
print(len(hometownDataSorted))
```

```python
AllData = []
for i, v in enumerate(repsDataSorted):
    AllData.append(v.copy())
    AllData[-1].append(hometownDataSorted[i][2]) #could repeat for committee count as well
AllData
```

```python
for row in AllData:
    row[2] = int(row[2])
    row[4] = int(row[4])
    row[3] = float(row[3].replace('$',"").replace(',',"")) #converting the money spent to a float by rmoving $ and ,
AllData
```

```python
moneySpentOverDays = []
for row in AllData:
    moneySpentOverDays.append(row[3]/row[2])
```

```python
miles = np.array(AllData[:2])
MoneySpentOverDaysArray = np.array(moneySpentOverDays[:0])

plt.plot(miles, MoneySpentOverDaysArray, 'm^')
```
